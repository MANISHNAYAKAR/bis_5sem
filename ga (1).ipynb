```python
import random

# Job Scheduling using Genetic Algorithm

# Jobs and their processing times
jobs = {
    'J1': 4,
    'J2': 2,
    'J3': 6,
    'J4': 3,
    'J5': 5
}

# Genetic Algorithm parameters
POPULATION_SIZE = 10
CROSSOVER_RATE = 0.8
MUTATION_RATE = 0.2
GENERATIONS = 50

# Fitness function: Minimize total completion time
def fitness(schedule):
    completion_time = 0
    total_completion_time = 0

    for job in schedule:
        completion_time += jobs[job]
        total_completion_time += completion_time

    return 1 / (total_completion_time + 1)

# Generate a random schedule
def create_individual():
    schedule = list(jobs.keys())
    random.shuffle(schedule)
    return schedule

# Create initial population
def create_population():
    return [create_individual() for _ in range(POPULATION_SIZE)]

# Selection: Choose the best two individuals
def selection(population):
    population.sort(key=fitness, reverse=True)
    return population[0], population[1]

# Crossover: Order crossover
def crossover(parent1, parent2):
    if random.random() > CROSSOVER_RATE:
        return parent1[:]

    start, end = sorted(random.sample(range(len(parent1)), 2))
    child = [None] * len(parent1)

    child[start:end + 1] = parent1[start:end + 1]
    remaining = [job for job in parent2 if job not in child]

    index = 0
    for i in range(len(child)):
        if child[i] is None:
            child[i] = remaining[index]
            index += 1

    return child

# Mutation: Swap two jobs
def mutation(schedule):
    if random.random() < MUTATION_RATE:
        i, j = random.sample(range(len(schedule)), 2)
        schedule[i], schedule[j] = schedule[j], schedule[i]
    return schedule

# Main Genetic Algorithm
population = create_population()

best_schedule = None
best_fitness = -1

for generation in range(GENERATIONS):
    population.sort(key=fitness, reverse=True)

    if fitness(population[0]) > best_fitness:
        best_schedule = population[0][:]
        best_fitness = fitness(best_schedule)

    new_population = [population[0][:]]  # Elitism

    while len(new_population) < POPULATION_SIZE:
        parent1, parent2 = selection(population)
        child = crossover(parent1, parent2)
        child = mutation(child)
        new_population.append(child)

    population = new_population

# Display the best schedule
print("Best Job Schedule:")
completion_time = 0
total_completion_time = 0

for job in best_schedule:
    completion_time += jobs[job]
    total_completion_time += completion_time
    print(f"{job}: Processing Time = {jobs[job]}, "
          f"Completion Time = {completion_time}")

print("\nTotal Completion Time:", total_completion_time)
print("Best Fitness Value:", round(best_fitness, 6))
```
